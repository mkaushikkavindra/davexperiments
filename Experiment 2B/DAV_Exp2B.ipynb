```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("googleplaystore.csv")

print("First 5 rows:\n")
print(df.head())

print("\nLast 5 rows:\n")
print(df.tail())

print("\nInformation:\n")
df.info()

print("\nSummary Statistics:\n")
print(df.describe(include="all"))
```

```python
# Fill missing ratings with mean
if "Rating" in df.columns:
    df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
    df["Rating"].fillna(df["Rating"].mean(), inplace=True)

print(df.isnull().sum())
```

```python
if "Rating" in df.columns:
    df["New_Rating"] = df["Rating"] * 2

print(df[["Rating","New_Rating"]].head())
```

```python
series = df["Rating"]

print(series.head())

print(series + 1)
```

```python
filtered_df = df[
    (df["Rating"] > 4.5) &
    (df["Type"] == "Free")
]

print(filtered_df.head())
```

```python
grouped = df.groupby("Category")["Rating"].mean()

print(grouped)
```

```python
sorted_df = df.sort_values(
    by="Rating",
    ascending=False
)

print(sorted_df.head())
```

```python
masked_df = df[df["Rating"] > df["Rating"].median()]

print(masked_df.head())
```

```python
df.drop_duplicates(inplace=True)

df.dropna(inplace=True)

print(df.shape)
```

```python
subset_df = df[["App","Category"]]

print(subset_df.head())
```

```python
subset_df.to_csv(
    "filtered_data.csv",
    index=False
)

print("CSV Saved")
```
