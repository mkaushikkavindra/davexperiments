```python
import pandas as pd

# Create CSV
csv_df = pd.DataFrame({
    "Product": ["Laptop", "Smartphone", "Tablet", "Headphones"],
    "Price": [1000, 800, 500, 100],
    "Quantity": [5, 8, 10, 15]
})
csv_df.to_csv("Google_data.csv", index=False)

# Create Excel
csv_df.to_excel("data.xlsx", index=False)

print("CSV and Excel files created successfully.")
```

```python
# Read CSV
text_df = pd.read_csv("Google_data.csv")

# Read Excel
excel_df = pd.read_excel("data.xlsx")

# Read Web Data
web_df = pd.read_csv("https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv")

print("CSV Data")
print(text_df.head())

print("\nExcel Data")
print(excel_df.head())

print("\nWeb Data")
print(web_df.head())
```

```python
# Handle missing values
text_df = text_df.ffill()
excel_df = excel_df.bfill()
web_df = web_df.dropna()

# Save processed files
text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("Processed files saved successfully.")
```
