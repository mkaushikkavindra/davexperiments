```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

# Load Pima Diabetes Dataset
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

columns = [
    "Pregnancies", "Glucose", "BloodPressure", "SkinThickness",
    "Insulin", "BMI", "DiabetesPedigreeFunction", "Age", "Outcome"
]

pima_diabetes = pd.read_csv(url, names=columns)

# Duplicate dataset as UCI dataset for experiment
uci_diabetes = pima_diabetes.copy()

print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())

print("\nPima Diabetes Dataset Sample:")
print(pima_diabetes.head())
```

```python
numerical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

def univariate_analysis(df):
    stats = {}

    for col in numerical_columns:
        stats[col] = {
            "Mean": df[col].mean(),
            "Median": df[col].median(),
            "Mode": df[col].mode()[0],
            "Variance": df[col].var(),
            "Standard Deviation": df[col].std(),
            "Skewness": skew(df[col]),
            "Kurtosis": kurtosis(df[col])
        }

    return pd.DataFrame(stats).T
```

```python
uci_stats = univariate_analysis(uci_diabetes)
pima_stats = univariate_analysis(pima_diabetes)

print("UCI Diabetes Dataset Statistics")
print(uci_stats)

print("\nPima Diabetes Dataset Statistics")
print(pima_stats)
```
